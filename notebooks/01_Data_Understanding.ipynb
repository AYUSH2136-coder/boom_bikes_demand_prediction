# 02 — Data Understanding

Inspect the raw dataset: shape, types, missing values, duplicates, and descriptive statistics.


```python
import pandas as pd
import numpy as np
import yaml
import warnings
warnings.filterwarnings('ignore')

# ── Load config ─────────────────────────────────────────────────────────────
with open('../configs/paths.yaml') as f:
    paths = yaml.safe_load(f)

RAW_DATA = paths['data']['raw']
print('Dataset path:', RAW_DATA)
```

```python
# ── Load dataset ─────────────────────────────────────────────────────────────
df = pd.read_csv(f'../{RAW_DATA}')

print(f'Shape: {df.shape}')
df.head()
```

```python
# ── Data types ───────────────────────────────────────────────────────────────
df.info()
```

```python
# ── Descriptive statistics ───────────────────────────────────────────────────
df.describe().T.style.background_gradient(cmap='Blues')
```

```python
# ── Missing values ───────────────────────────────────────────────────────────
missing = df.isnull().sum()
missing_pct = (missing / len(df)) * 100
pd.DataFrame({'missing_count': missing, 'missing_pct': missing_pct}).query('missing_count > 0')
```

```python
# ── Duplicate rows ────────────────────────────────────────────────────────────
print(f'Duplicate rows: {df.duplicated().sum()}')
```

```python
# ── Target distribution ───────────────────────────────────────────────────────
print('Target (cnt) summary:')
print(df['cnt'].describe())
print(f'\nSkewness: {df["cnt"].skew():.3f}')
print(f'Kurtosis: {df["cnt"].kurtosis():.3f}')
```

```python
# ── Categorical value counts ──────────────────────────────────────────────────
cat_cols = ['season', 'yr', 'mnth', 'holiday', 'weekday', 'workingday', 'weathersit']

for col in cat_cols:
    print(f'\n── {col} ──')
    print(df[col].value_counts().sort_index())
```

```python
# ── Date range ────────────────────────────────────────────────────────────────
df['dteday'] = pd.to_datetime(df['dteday'], dayfirst=True)
print(f'Date range: {df["dteday"].min().date()}  →  {df["dteday"].max().date()}')
print(f'Total days covered: {(df["dteday"].max() - df["dteday"].min()).days + 1}')
print(f'Rows in dataset:    {len(df)}')

# Check for gaps
all_dates = pd.date_range(df['dteday'].min(), df['dteday'].max(), freq='D')
missing_dates = set(all_dates) - set(df['dteday'])
print(f'Missing dates in time series: {len(missing_dates)}')
```
